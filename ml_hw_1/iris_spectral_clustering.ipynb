# Спектральна кластеризація. Набір даних Iris

Імпортуємо потрібні бібліотеки

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import SpectralClustering
from sklearn.metrics import confusion_matrix
```

### 1. Завантажуємо дані і робимо DataFrame

```python
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target
df.head()
```

### 2. Статистика

```python
df.describe()
```

```python
df['target'].value_counts()
```

По 50 спостережень в кожному класі, дані збалансовані.

### 3. Дивимось як розподілені дані по класах

```python
sns.pairplot(df, hue='target')
plt.show()
```

Видно що один клас (setosa) добре відділяється, а два інших трохи перемішані.

### 4. Стандартизація

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(iris.data)
X_scaled[:5]
```

### 5. Спектральна кластеризація

```python
sc = SpectralClustering(n_clusters=3, random_state=42)
labels = sc.fit_predict(X_scaled)
labels
```

### 6. Порівнюємо з реальними класами

```python
cm = confusion_matrix(iris.target, labels)
cm
```

```python
sns.heatmap(cm, annot=True, cmap='Blues')
plt.xlabel('кластер')
plt.ylabel('реальний клас')
plt.show()
```

Номери кластерів не співпадають з номерами класів (бо алгоритм їх присвоює довільно), але видно що кожен клас в основному потрапив в свій окремий кластер.

### 7. Візуалізація результату

```python
plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
plt.scatter(iris.data[:,2], iris.data[:,3], c=iris.target)
plt.title('реальні класи')

plt.subplot(1,2,2)
plt.scatter(iris.data[:,2], iris.data[:,3], c=labels)
plt.title('кластери')

plt.show()
```

### Висновок

В цій роботі я зробила спектральну кластеризацію на датасеті Iris.

Спочатку завантажила дані, подивилась статистику і графіки. Дані виявились збалансованими, по 50 в кожному класі. По графіках видно що клас setosa легко відділяється, а versicolor і virginica частково перемішані.

Перед кластеризацією зробила стандартизацію, бо ознаки в різних масштабах. Потім застосувала SpectralClustering на 3 кластери.

По confusion matrix видно що кластеризація вийшла досить непогана - setosa повністю в одному кластері, а основні помилки між versicolor і virginica, бо вони схожі між собою. Тобто алгоритм майже правильно знайшов 3 групи без використання міток класів.
